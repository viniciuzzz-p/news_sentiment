```python
import re
import pandas as pd
```

```python
df = pd.read_csv('../data/raw/mastodon.csv')
```

```python
df["text_clean"] = df['text']
```

### Todo texto em minúsculo

```python
df["text_clean"] = df["text_clean"].str.lower()
df
```

### Remover hhtp

```python
df["text_clean"] = df["text_clean"].apply(
    lambda x: re.sub(r"http\S+", "", x)
)
```

### Remover menções (@user)

```python
df["text_clean"] = df["text_clean"].apply(
    lambda x: re.sub(r"@\w+", "", x)
)
```

### Remover hashtags

```python
df["text_clean"] = df["text_clean"].str.replace("#", "", regex=False)
```

### Remover espaços extras

```python
df["text_clean"] = df["text_clean"].apply(
    lambda x: re.sub(r"\s+", " ", x).strip()
)
```

```python
df.to_csv('../data/raw/mastodon.csv', index= False)
```
