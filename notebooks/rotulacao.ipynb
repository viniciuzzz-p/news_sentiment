```python
import pandas as pd
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
analyzer = SentimentIntensityAnalyzer()
```

```python
df = pd.read_csv('../data/raw/mastodon.csv')
df
```

```python
#embaralha as linhas
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
sentiment_score = []
```

```python
for text in df['text_clean']:
    text = str(text)

    scores = analyzer.polarity_scores(text)
    compound_score = scores["compound"]

    sentiment_score.append(compound_score)
```

```python
df["sentiment_score"] = sentiment_score
```

```python
labels = []
```

```python
for score in df['sentiment_score']:
    if score > 0.05:
        labels.append(1) #comentario positivo
    
    elif score < -0.05:
        labels.append(-1) #comentario negativo
    
    else:
        labels.append(0) #comentario neutro
    

```

```python
df['labels'] = labels
```

```python
df = df.dropna()
df
```
